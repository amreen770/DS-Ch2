## Whitespace formatting:
Whitespace means spaces or tabs used before a line of code. Python uses identation. It is ignored inside parentheses and brackets which is very useful for long computations.


```python
for i in [1, 2, 3, 4, 4]:
    print i
    for j in [1, 2, 3, 4, 5]:
        print j
        print i+j
    print i
print "done looping"    
```
